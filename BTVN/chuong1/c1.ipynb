```python
class HinhChuNhat:
    def __init__(self, dai=0, rong=0):
        self.dai = dai
        self.rong = rong

    # Phương thức nhập dữ liệu
    def nhap(self):
        self.dai = float(input("Nhập chiều dài: "))
        self.rong = float(input("Nhập chiều rộng: "))

    # Tính chu vi
    def chu_vi(self):
        return 2 * (self.dai + self.rong)

    # Tính diện tích
    def dien_tich(self):
        return self.dai * self.rong

    # In thông tin
    def hien_thi(self):
        print("----- Thông tin hình chữ nhật -----")
        print(f"Chiều dài: {self.dai}")
        print(f"Chiều rộng: {self.rong}")
        print(f"Chu vi: {self.chu_vi()}")
        print(f"Diện tích: {self.dien_tich()}")


# Chương trình chính
if __name__ == "__main__":
    hcn = HinhChuNhat()
    hcn.nhap()
    hcn.hien_thi()
```
