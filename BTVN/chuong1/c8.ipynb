```python
class Date:
    def __init__(self, ngay=1, thang=1, nam=2000):
        self.ngay = ngay
        self.thang = thang
        self.nam = nam

    def display(self):
        return f"{self.ngay}/{self.thang}/{self.nam}"

class Employee:
    def __init__(self, ho_ten, ngay_sinh, ngay_vao_ct):
        self.ho_ten = ho_ten
        self.ngay_sinh = ngay_sinh      
        self.ngay_vao_ct = ngay_vao_ct  

    def display(self):
        print(f"Họ tên: {self.ho_ten}")
        print(f"Ngày sinh: {self.ngay_sinh.display()}")
        print(f"Ngày vào công ty: {self.ngay_vao_ct.display()}")

if __name__ == "__main__":
    nv1 = Employee("Nguyễn Văn A", Date(12, 5, 1999), Date(1, 1, 2020))
    nv2 = Employee("Trần Thị B", Date(23, 8, 2000), Date(15, 7, 2021))
    print("===== DANH SÁCH NHÂN VIÊN =====")
    nv1.display()
    print("-----------------------------")
    nv2.display()
```
