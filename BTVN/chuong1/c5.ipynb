```python
class NganXep:
    def __init__(self, kich_thuoc):
        self.kich_thuoc = kich_thuoc           # dung lượng tối đa
        self.mang = [0.0] * kich_thuoc         # mảng float
        self.dinh = -1                         # vị trí đỉnh ngăn xếp

    def __del__(self):  
        print(" Ngăn xếp đã được hủy.")

    def rong(self):
        return self.dinh == -1

    def day(self):
        return self.dinh == self.kich_thuoc - 1

    def them(self, gia_tri):
        if self.day():
            print(" Ngăn xếp đã đầy, không thể thêm!")
        else:
            self.dinh += 1
            self.mang[self.dinh] = float(gia_tri)
            print(f" Đã thêm {gia_tri} vào ngăn xếp.")


    def lay(self):
        if self.rong():
            print(" Ngăn xếp rỗng, không thể lấy!")
            return None
        else:
            gia_tri = self.mang[self.dinh]
            self.dinh -= 1
            print(f"Đã lấy {gia_tri} ra khỏi ngăn xếp.")
            return gia_tri

   
    def hien_thi(self):
        if self.rong():
            print(" Ngăn xếp hiện tại: RỖNG")
        else:
            print(" Ngăn xếp (từ dưới lên): ", end="")
            for i in range(self.dinh + 1):
                print(self.mang[i], end=" ")
            print()

    # Đếm số phần tử hiện tại 
    def dem(self):
        return self.dinh + 1   # vì self.dinh bắt đầu từ -1
# main
if __name__ == "__main__":
    n = int(input("Nhập kích thước tối đa của ngăn xếp: "))
    ngan_xep = NganXep(n)

    ngan_xep.them(10)
    ngan_xep.them(20)
    ngan_xep.them(30)
    ngan_xep.hien_thi()

    print(f" Số phần tử hiện tại trong ngăn xếp: {ngan_xep.dem()}")

    ngan_xep.lay()
    ngan_xep.hien_thi()
    print(f"Số phần tử hiện tại trong ngăn xếp: {ngan_xep.dem()}")
```
