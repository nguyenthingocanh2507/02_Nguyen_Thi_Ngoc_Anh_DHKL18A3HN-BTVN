```python
class Date:
    def __init__(self, ngay=1, thang=1, nam=2000):
        self.ngay = ngay
        self.thang = thang
        self.nam = nam

    def nam_nhuan(self, nam):
        return (nam % 400 == 0) or (nam % 4 == 0 and nam % 100 != 0)

    def so_ngay_trong_thang(self, thang, nam):
        if thang in [1, 3, 5, 7, 8, 10, 12]:
            return 31
        elif thang in [4, 6, 9, 11]:
            return 30
        elif thang == 2:
            return 29 if self.nam_nhuan(nam) else 28
        else:
            return 0

    def hien_thi(self):
        print(f"{self.ngay:02d}/{self.thang:02d}/{self.nam}")

    def ngay_tiep(self):
        ngay_max = self.so_ngay_trong_thang(self.thang, self.nam)
        if self.ngay < ngay_max:
            self.ngay += 1
        else:
            self.ngay = 1
            if self.thang == 12:
                self.thang = 1
                self.nam += 1
            else:
                self.thang += 1

if __name__ == "__main__":
    d = Date(28, 2, 2024)   # 28/02/2024 (năm nhuận)
    print("Ngày hiện tại:", end=" ")
    d.hien_thi()

    d.ngay_tiep()
    print("Ngày tiếp theo:", end=" ")
    d.hien_thi()

    d.ngay_tiep()
    print("Ngày tiếp theo:", end=" ")
    d.hien_thi()
```
