```python
class PhanSo:
    def __init__(self, tu=0, mau=1):
        self.tu = tu
        self.mau = mau
        self.hop_le = self.kiem_tra_hop_le()

    # Kiểm tra tính hợp lệ (mẫu số ≠ 0)
    def kiem_tra_hop_le(self):
        if self.mau == 0:
            return False
        return True

    # Nhập phân số
    def nhap(self):
        self.tu = int(input("Nhập tử số: "))
        self.mau = int(input("Nhập mẫu số: "))
        while self.mau == 0:
            print("❌ Mẫu số không được bằng 0, vui lòng nhập lại!")
            self.mau = int(input("Nhập mẫu số: "))
        self.hop_le = True

    # In phân số
    def hien_thi(self):
        if not self.hop_le:
            print("Phân số không hợp lệ (mẫu số = 0)")
        else:
            if self.mau == 1:
                print(f"Phân số: {self.tu}")   # nếu mẫu = 1 thì chỉ in tử số
            else:
                print(f"Phân số: {self.tu}/{self.mau}")


# =========================
# Chương trình chính
# =========================
if __name__ == "__main__":
    ps = PhanSo()
    ps.nhap()
    ps.hien_thi()
```
