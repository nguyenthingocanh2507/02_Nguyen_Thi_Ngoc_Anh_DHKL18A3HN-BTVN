```python
class ThiSinh:
    def __init__(self, ho_ten="", toan=0, ly=0, hoa=0):
        self.ho_ten = ho_ten
        self.toan = toan
        self.ly = ly
        self.hoa = hoa

    # Nhập thông tin thí sinh
    def nhap(self):
        self.ho_ten = input("Nhập họ tên thí sinh: ")
        self.toan = float(input("Nhập điểm Toán: "))
        self.ly = float(input("Nhập điểm Lý: "))
        self.hoa = float(input("Nhập điểm Hóa: "))

    # Tính tổng điểm
    def tong_diem(self):
        return self.toan + self.ly + self.hoa

    # In thông tin thí sinh
    def hien_thi(self):
        print(f"Họ tên: {self.ho_ten}")
        print(f"Điểm Toán: {self.toan}, Lý: {self.ly}, Hóa: {self.hoa}")
        print(f"Tổng điểm: {self.tong_diem()}")
        print("-" * 30)


# ======================
# Chương trình chính
# ======================
if __name__ == "__main__":
    ds_thi_sinh = []
    n = int(input("Nhập số lượng thí sinh: "))

    # Nhập danh sách thí sinh
    for i in range(n):
        print(f"\nNhập thông tin thí sinh thứ {i+1}:")
        ts = ThiSinh()
        ts.nhap()
        ds_thi_sinh.append(ts)

    # Lọc danh sách trúng tuyển (điểm chuẩn 20)
    diem_chuan = 20
    ds_trung_tuyen = [ts for ts in ds_thi_sinh if ts.tong_diem() >= diem_chuan]

    # Sắp xếp theo tổng điểm giảm dần
    ds_trung_tuyen.sort(key=lambda x: x.tong_diem(), reverse=True)

    # In kết quả
    print("\n===== DANH SÁCH THÍ SINH TRÚNG TUYỂN =====")
    if not ds_trung_tuyen:
        print("Không có thí sinh nào trúng tuyển.")
    else:
        for ts in ds_trung_tuyen:
            ts.hien_thi()
```
