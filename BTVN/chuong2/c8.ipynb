```python
import json
doi_tuong_python = {
    "ten": "Nguyen Van C",
    "tuoi": 22,
    "lop": "11A4",
    "diem": [7.5, 8.0, 9.0]
}

json_str = json.dumps(doi_tuong_python)

print("Chuỗi JSON từ đối tượng Python:")
print(json_str)
print("\nCác giá trị của đối tượng:")
for key, value in doi_tuong_python.items():
    print(f"{key}: {value}")
```
