```python
import sqlite3

db_path = "example.db"

with sqlite3.connect(db_path) as conn:
    cur = conn.cursor()

    cur.execute("DROP TABLE IF EXISTS sample")

    cur.execute("""
    CREATE TABLE sample (
        id INTEGER PRIMARY KEY AUTOINCREMENT,
        name TEXT,
        age INTEGER
    )
    """)

    nhan_vien = [
        ("Nguyen Van A", 30),
        ("Tran Thi B", 25),
        ("Le Van C", 28)
    ]

    cur.executemany(
        "INSERT INTO sample (name, age) VALUES (?, ?)",
        nhan_vien
    )

    liet_ke = cur.execute("SELECT * FROM sample").fetchall()

    for row in liet_ke:
        print(row)

    conn.commit()
```

```python
import sqlite3

with sqlite3.connect("product.db") as conn:
    cur = conn.cursor()

    # Tạo bảng nếu chưa có
    cur.execute("""
    CREATE TABLE IF NOT EXISTS products (
        id INTEGER PRIMARY KEY AUTOINCREMENT,
        product_name TEXT,
        price INTEGER
    )
    """)

    # Thêm dữ liệu
    san_pham = [
        ("Sản phẩm A", 10000),
        ("Sản phẩm B", 20000),
        ("Sản phẩm C", 15000)
    ]

    cur.executemany(
        "INSERT INTO products (product_name, price) VALUES (?, ?)",
        san_pham
    )

    # Hiển thị danh sách
    liet_ke = cur.execute("SELECT * FROM products").fetchall()
    print("Hiển thị danh sách sản phẩm:", liet_ke)

    # Nhập ID cần tìm
    id = int(input("Nhập ID sản phẩm cần tìm kiếm: "))

    # Tìm kiếm
    tim_kiem = cur.execute(
        "SELECT * FROM products WHERE id = ?",
        (id,)
    ).fetchall()
    print("Kết quả tìm kiếm:", tim_kiem)

    # Cập nhật giá
    cur.execute(
        "UPDATE products SET price = price + 5000 WHERE id = ?",
        (id,)
    )

    # Xóa sản phẩm
    cur.execute(
        "DELETE FROM products WHERE id = ?",
        (id,)
    )

    conn.commit()
```
