```python
#chương 3 - NUMPY - thư viện nền tảng của python \
# thiết kế chuyên biệt xử lý dữ liệu số học ,đstt,mảng đa chiều\
# thao tác trên các tập dữ liệu lớn 
# đều cùng 1 kiểu dlieu khog giống List ( string ,float)

import numpy as np

import sys
# tạo 1 ds python với 1000 ptu
S = list(range(1000))

# đo kthuoc mỗi ptu trog list
item_s= sys.getsizeof(S)
print("kích thước của mỗi phần tử trong list tính theo bytes ",item_s)

# đo kích thước của toàn bộ ds S
S_size = item_s*len(S)
print("kích thước của toàn bộ list ....",S_size)

# tạo 1 mảng numpy array với 1000 ptu
D = np.arange(1000)
# đo kthuoc của mỗi ptu trog numpy
print ("kthuoc của mỗi ptu của numpy array",D.itemsize)
# ĐO KTHUOC CỦA TOÀN bộ
array_size= D.itemsize*D.size
print("kích thước toàn bộ mảng numpy",array_size)

print(f"tỉ lệ (list/array):,{S_size / array_size:.2f}")


# IN RA TA THẤY NUMPY SẼ CÓ KTHUOC > 1000 LẦN VỚI LIST
```
