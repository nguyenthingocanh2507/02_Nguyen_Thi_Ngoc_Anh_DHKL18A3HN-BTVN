```python
import json 
# Import lớp JSONDecoder từ module json.decoder 
from json.decoder import JSONDecoder 
 
# Chuỗi JSON chứa dữ liệu màu sắc. 
colour_string = '{ "colour": ["red", "yellow"]}' 
 
# Sử dụng JSONDecoder để giải mã chuỗi JSON thành đối tượng Python. 
# JSONDecoder là một lớp con của json.JSONDecoder, cho phép tùy chỉnh quá trình 
#giải mã. 
# Phương thức decode() sẽ giải mã chuỗi JSON và trả về đối tượng Python tương ứng
decoded_object = JSONDecoder().decode(colour_string) 
 
# In đối tượng Python đã giải mã và kiểu của đối tượng đó ra màn hình. 
print(decoded_object) 
print(type(decoded_object))
```

```python
import json 
 
# Hàm kiểm tra chuỗi JSON có chứa đối tượng phức (complex object) hay không 
def is_complex(object): 
    # Kiểm tra xem trong đối tượng có tồn tại khóa '__complex__' hay không 
    if '__complex__' in object: 
        # Nếu có, tạo một số phức từ thông tin trong đối tượng 
        return complex(object['real'], object['img']) 
    # Nếu không có khóa '__complex__', đối tượng không phải là đối tượng phức,  
          # trả về đối tượng gốc 
    return object 
 
def main(): 
    # Sử dụng hàm json.loads() với object_hook để kiểm tra và chuyển đổi đối tượng       
          #phức trong chuỗi JSON 
    complex_object = json.loads('{"__complex__": true, "real": 4, "img": 5}',   
                               object_hook=is_complex) 
    simple_object = json.loads('{"real": 6, "img": 7}',  
                                object_hook=is_complex) 
 
    # In kết quả phân tích chuỗi JSON cho cả đối tượng phức và đối tượng thường 
    print("Đối tượng số phức......", complex_object) 
    print("Đối tượng bình thường......", simple_object) 
 
if __name__ == "__main__": 
    main()
```

```python
# default:mặc định là none , cho phép phthuc xử lý các đối tượng  k thể mã hoá json
import json

def complex_handler(obj): # hàm xử lý số phức ktra nếu có obj là số phức thì trả về ds gồm phần thực và phần ảo 
    if isinstance(obj,complex): 
        return [obj.real,obj.imag] # trả về ds gồm phần thực và phần ảo của số phức
    raise TypeError(repr(obj) +"k thể mã hoá " ) # NẾU kp số phức thì ném về lỗi typ..
# ứng dụng hàm comple handler để mã hoá số phức 
data={"z":2+3j}
complex_obj= json.dumps(data,default=complex_handler) # sử dụng đối số default để chỉ định hàm xử lý tuỳ chỉnh cái complex hanlder
print(complex_obj)

```

```python
# chuyển đổi giữa đối tượng python <-> chuỗi json
#serialize (tuần tự hoá ): chuyển đối tượng python -> json
# deserialize ( giải tuần tự hoá ): chuyển ngược lại 
# module json hỗ trợ đầy đủ 2 quá  trình này 
# json.dumps() : chuyển python -> json
# json.dump(): chuyển json -> python
import json

# Tạo đối tượng Python (kiểu dict)
data = {
    "name": "Ng Văn A",
    "age": 23,
    "skills": ["AI", "Python"]
}
# Chuyển đối tượng Python sang chuỗi JSON, đối số ensure để giữ nguyên ký hiệu 
json_st1= json.dumps(data, ensure_ascii=False)
# In ra chuỗi JSON
print(data) 
print(json_st1)  
# indent định dạng về các dòng lưu trữ 
json_st2=json.dumps(data,indent=4,ensure_ascii=False)
print(json_st2)
```
