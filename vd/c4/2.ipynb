```python
# pandas index 
# index là đối tượng bất biến - đóng vtro nhãn chính cho các trục
import pandas as pd
ind=pd.Index([2,3,6,7,11]) # tạo index 
ind
# cắt lát theo nhãn ' .loc '
# cắt lát theo vị trí ' .iloc ' 
# vdu truy cập và cắt lát 
print('ind[1] =',ind[1]) # truy cập vào phần tử thứ 2 trog ind
print('ind[::2]=',ind[::2]) # cắt lát với bước nhảy 2
# thuộc tính tương tự numpy
print("size=",ind.size)
print("shape=",ind.shape)
print("ndim= ",ind.ndim)

try: # ktra lỗi 
    ind[3]=5
except Exception as e:
    print("lỗi khi gán phần tử index",type(e).__name__,'-',e)
```

```python
indA=pd.Index([0,2,4,6,8])
indB=pd.Index([1,3,7,8,9])

print('index a',indA)
print("index b",indB)
print('\n hợp (union):',indA.union(indB))
print('hiệu (difference a - b ):',indA.difference(indB))
print('giao (difference):',indA.intersection(indB))
print("hiệu đối xứng (symmetric_difference):",indA.symmetric_difference(indB)) 
print('\nA thuộc B? (isin):',indA.isin(indB)) # KTRA phần tử (isin)
```
