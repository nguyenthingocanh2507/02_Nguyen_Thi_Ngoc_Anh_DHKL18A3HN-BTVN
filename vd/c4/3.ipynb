```python
import pandas as pd
index_tuples ={
    ('Ha NOi', 2015), ('Ha Noi', 2020),
    ('Nam Dinh', 2015), ('Nam Dinh', 2020),
    ('Thanh Hoa', 2015), ('Thanh Hoa', 2020),
    ('Da Nang', 2015), ('Da Nang', 2020),
    ('Ho Chi Minh', 2015), ('Ho Chi Minh', 2020)
}
data = [
    7216000, 8246548,
    1850600, 1780330,
    3502000, 3603000,
    1125400, 1156000,
    8421000, 9153000
]
vn_pop = pd.Series(data, index=index_tuples)
print("Series su dung Index bangw Tuple ( Cach thu cong):")
vn_pop
# ycau 1 cho biet dân số HN (2015,2020) và nđịnh 2020
vn_pop = vn_pop.sort_index()   # hoặc vn_pop.sort_index(inplace=True)
vn_pop.loc[('Hà Nội', 2015):('Nam Định', 2015)]

```

```python
# multipledex
#4.6.1.4
multi_idx= pd.MultiIndex.from_tuples(index_tuples,names=['Tỉnh/TP','Năm'])
multi_idx
# reindex series theo multilndex vừa tạo
vn_pop_mul= vn_pop.reindex(multi_idx) # reindex lại seri vn_pop theo multinlndex mới là multi_idx, kqua trả vào 1 seri mới mul_mul
vn_pop_mul
# series - 1 chiều
# dataframe - 2 chiều 
type(vn_pop_mul)
# truy vấn theo cấp toàn bộ năm 2020 khi nó là multidex
vn_pop_mul[:,2020]
vn_pop[:,2020] # code này sẽ lỗi vì vn_pop có chỉ mục kp là multidex
# chỉ mục đa cấp theo chiều mở rộng- unstack(): chuuyeenr đổi 1 series có chỉ mục đa cấp thành 1 dataframe
# chuyển series thành dataframe
vn_pop_df= vn_pop_mul.unstack('Năm')
print("dataframe dạng rộng (cột là năm):")
display(vn_pop_df)
print("\nseries dạng dài")
display(vn_pop_df)
# sử dụng hàm pd.stack( để chuyển data thành series có multidex)
print("\nseries dạng dài(stack lại)")
display(vn_pop_df.stack())
type(vn_pop_df.stack()) # KTRA LẠi

```
